# <span style="color: #8ab0bc;">ライブラリインポート(Import Libraries)</span>

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

# <span style="color: #8ab0bc;">URL(Set Target URL)</span>

```python
url = "https://www.kkr.mlit.go.jp/road/michi_no_eki/ichiran.html#prefecture-4"
```

# <span style="color: #8ab0bc;">HTML抽出(Fetch HTML Content)</span>

```python
r = requests.get(url)
r.encoding = "utf-8"
r.text
```

```python
soup =  BeautifulSoup(r.text, 'html.parser')
soup
```

# <span style="color: #8ab0bc;">タグを指定(Parse HTML Elements)</span>

```python
table_body = soup.find_all("ul",class_="table body")
```

# <span style="color: #8ab0bc;">データ整理(Data Cleaning & Structuring)</span>

```python
contents_data = []
head_url = "https://www.kkr.mlit.go.jp/road/michi_no_eki/"

for i in range(len(table_body)):
    name = table_body[i].find("li", class_="station-name").text.strip()
    location = table_body[i].find("li", class_="location").text.strip()
    URL = head_url + table_body[i].find("a").get("href")
    

    data = {
        "駅名": name,
        "住所": location,
        "URL": URL       
    }
    
    contents_data.append(data)
```

```python
contents_data
```

# <span style="color: #8ab0bc;">PANDAS DF作成(Create Pandas DataFrame)</span>

```python
df = pd.DataFrame(contents_data)
df
```

# <span style="color: #8ab0bc;">CSVに出力(Export to CSV)</span>

```python
df.to_csv("mitinoeki.csv", index=False)
```
